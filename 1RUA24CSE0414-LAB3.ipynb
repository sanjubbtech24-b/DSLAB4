```python
df['emi']=((df['Emi_or_Rent_Amt'])/(df['Mthly_HH_Income']))*100
print(df['emi'].mean())
```

```python
df['new_minc'] = df['Mthly_HH_Income']*0.4
houseind=df[df['Emi_or_Rent_Amt'] > df['new_minc']]

print(houseind.index.tolist())
```

```python
df['lowdis']=df['Mthly_HH_Income']-df['Mthly_HH_Expense']-df['Emi_or_Rent_Amt']

idx = df['lowdis'].idxmin()
print(idx)
```

```python
df['expec_inc'] = df['Mthly_HH_Income']*12
dis = df['Annual_HH_Income'] - df['expec_inc']
print(dis)
```

```python
group = df.groupby('Highest_Qualified_Member')
avg_inc_by_qual_mean = group['Mthly_HH_Income'].mean()

avg_inc_by_qual_median = group['Mthly_HH_Income'].median()

avg_income_by_qual = pd.DataFrame({
    'mean': avg_inc_by_qual_mean,
    'median': avg_inc_by_qual_median
})

print(avg_income_by_qual)
```

```python
df['Zscore'] = stats.zscore(df['Mthly_HH_Income'])
outliers_inc = df[np.abs(df['Zscore']) > 3]
print(outliers_inc)
corr_inc_expense = df['Mthly_HH_Income'].corr(df['Mthly_HH_Expense'])
print(corr_inc_expense)
corr_earners_inc = df['No_of_Earning_Members'].corr(df['Mthly_HH_Income'])
print(corr_earners_inc)
```
