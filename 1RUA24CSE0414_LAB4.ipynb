```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    "Branch": ["A","C","A","A","A","C","A","C","A","B","B","B","A","A","A","B","A"],
    "City": ["Yangon","Naypyitaw","Yangon","Yangon","Yangon","Naypyitaw","Yangon","Naypyitaw",
             "Yangon","Mandalay","Mandalay","Mandalay","Yangon","Yangon","Yangon","Mandalay","Yangon"],
    "Customer type": ["Member","Normal","Normal","Member","Normal","Normal","Member","Normal",
                      "Member","Member","Member","Member","Normal","Normal","Normal","Member","Member"],
    "Gender": ["Female","Female","Male","Male","Male","Male","Female","Female","Female","Female",
               "Female","Male","Female","Male","Female","Female","Female"],
    "Unit price": [74.69,15.28,46.33,58.22,86.31,85.39,68.84,73.56,36.26,54.84,14.48,25.51,46.95,43.19,71.38,93.72,68.93],
    "Quantity": [7,5,7,8,7,7,6,10,2,3,4,4,5,10,10,6,7],
    "Tax 5%": [26.1415,3.82,16.2155,23.288,30.2085,29.8865,20.652,36.78,3.626,8.226,2.896,5.102,11.7375,21.595,35.69,28.116,24.1255],
    "Total": [548.9715,80.22,340.5255,489.048,634.3785,627.6165,433.692,772.38,76.146,172.746,60.816,107.142,246.4875,453.495,749.49,590.436,506.6355]
}
```

```python
df = pd.DataFrame(data)
df.head()
```

```python
plt.figure(figsize=(7, 4))
plt.plot(df["Unit price"], df["Total"], marker='o', color='b')
plt.title("Line Plot: Unit Price vs Total")
plt.xlabel("Unit Price")
plt.ylabel("Total")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.scatterplot(x="Unit price", y="Total", data=df, hue="Gender", style="Customer type", s=80)
plt.title("Scatter Plot: Unit Price vs Total")
plt.xlabel("Unit Price")
plt.ylabel("Total")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.regplot(x="Unit price", y="Quantity", data=df, scatter_kws={"s": 60}, line_kws={"color": "red"})
plt.title("Linear Regression Plot: Unit Price vs Quantity")
plt.xlabel("Unit Price")
plt.ylabel("Quantity")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(x="Gender", data=df, palette="pastel")
plt.title("Bar Plot: Number of Male and Female Customers")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(x="Customer type", data=df, palette="Set2")
plt.title("Bar Plot: Number of Member and Normal Customers")
plt.xlabel("Customer Type")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
avg_price = df.groupby("Gender")["Unit price"].mean().reset_index()
plt.figure(figsize=(6, 4))
sns.barplot(x="Gender", y="Unit price", data=avg_price, palette="coolwarm")
plt.title("Bar Plot: Average Unit Price by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Unit Price")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.boxplot(x="City", y="Unit price", data=df, palette="husl")
plt.title("Box Plot: City-wise Unit Price Distribution")
plt.xlabel("City")
plt.ylabel("Unit Price")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.swarmplot(x="Customer type", y="Quantity", data=df, hue="Gender", palette="Set1", size=7)
plt.title("Swarm Plot: Customer Type vs Quantity")
plt.xlabel("Customer Type")
plt.ylabel("Quantity")
plt.legend(title="Gender")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.violinplot(x="Customer type", y="Quantity", data=df, inner="quartile", palette="muted")
plt.title("Violin Plot: Customer Type vs Quantity")
plt.xlabel("Customer Type")
plt.ylabel("Quantity")
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.swarmplot(ax=axes[0], x="Customer type", y="Quantity", data=df, hue="Gender", palette="Set2", size=6)
axes[0].set_title("Swarm Plot: Customer Type vs Quantity")
axes[0].set_xlabel("Customer Type")
axes[0].set_ylabel("Quantity")
axes[0].legend(title="Gender")

sns.violinplot(ax=axes[1], x="Customer type", y="Quantity", data=df, inner="quartile", palette="cool")
axes[1].set_title("Violin Plot: Customer Type vs Quantity")
axes[1].set_xlabel("Customer Type")
axes[1].set_ylabel("Quantity")

plt.tight_layout()
plt.show()
```
